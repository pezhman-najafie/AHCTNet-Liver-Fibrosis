# Spatial-Frequency

```python
# =====================================================
# IMPORTS
# =====================================================
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    roc_curve,
    precision_score,
    recall_score,
    matthews_corrcoef,
    cohen_kappa_score,
    balanced_accuracy_score
)

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader

import torchvision.transforms as transforms
import torchvision.models as models

import seaborn as sns

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# =====================================================
# SPATIAL-FREQUENCY CONFIGURATION
# =====================================================
INPUT_SIZE = 256
OUTPUT_SIZE = 300
FREQUENCY_BORDER = OUTPUT_SIZE - INPUT_SIZE  # 44 pixels

assert FREQUENCY_BORDER > 0, (
    "OUTPUT_SIZE must be larger than INPUT_SIZE."
)

# =====================================================
# OUTPUT PATHS
# =====================================================
OUTPUT_DIR = "training_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

EPOCH_METRICS_CSV = os.path.join(
    OUTPUT_DIR,
    "epoch_metrics.csv"
)

BEST_OVERALL_MODEL_PATH = os.path.join(
    OUTPUT_DIR,
    "best_model_overall.pth"
)

BEST_OVERALL_INFO_CSV = os.path.join(
    OUTPUT_DIR,
    "best_model_overall_info.csv"
)


# =====================================================
# DATASET
# =====================================================
class LiverDataset(Dataset):

    def __init__(self, samples, transform=None):

        self.samples = samples
        self.transform = transform

    def __len__(self):

        return len(self.samples)

    def __getitem__(self, idx):

        path, label = self.samples[idx]

        img = Image.open(path).convert("RGB")

        if self.transform:
            img = self.transform(img)

        return img, label


# =====================================================
# LOAD DATASET
# =====================================================
def load_dataset(root):

    samples = []

    classes = sorted(os.listdir(root))

    for i, cls in enumerate(classes):

        cls_path = os.path.join(root, cls)

        for img in os.listdir(cls_path):

            samples.append(
                (
                    os.path.join(cls_path, img),
                    i
                )
            )

    return samples


# =====================================================
# TRANSFORMS
# =====================================================
train_tf = transforms.Compose([

    transforms.Resize((INPUT_SIZE, INPUT_SIZE)),

    transforms.RandomHorizontalFlip(),

    transforms.RandomRotation(20),

    transforms.ColorJitter(
        0.2,
        0.2,
        0.2,
        0.1
    ),

    transforms.ToTensor(),
])

val_tf = transforms.Compose([

    transforms.Resize((INPUT_SIZE, INPUT_SIZE)),

    transforms.ToTensor(),
])


# =====================================================
# FOCAL LOSS
# =====================================================
class FocalLoss(nn.Module):

    def __init__(
        self,
        gamma=2,
        smoothing=0.1
    ):

        super().__init__()

        self.gamma = gamma
        self.smoothing = smoothing

    def forward(self, logits, targets):

        ce = F.cross_entropy(
            logits,
            targets,
            reduction='none',
            label_smoothing=self.smoothing
        )

        pt = torch.exp(-ce)

        loss = (
            (1 - pt) ** self.gamma * ce
        ).mean()

        return loss


# =====================================================
# LEARNABLE SPATIAL-FREQUENCY AUGMENTATION
# =====================================================
class LearnableSpatialFrequencyAugmentation(nn.Module):
    """
    Converts a BxCx256x256 image batch to BxCx300x300 by combining:

      - the original spatial image weighted by alpha_image
      - FFT magnitude information weighted by alpha_frequency

    The two weights are learned jointly with the network and are constrained
    by softmax, therefore:

        alpha_image + alpha_frequency = 1
        alpha_image, alpha_frequency > 0

    Layout of the 300x300 tensor:

        +------------------------+----------+
        |                        |          |
        |  alpha_image * image   |  beta*F  |  256 rows
        |       256 x 256        | 256 x 44 |
        |                        |          |
        +------------------------+----------+
        | beta*F 44 x 256        | beta*F   |  44 rows
        |                        | 44 x 44  |
        +------------------------+----------+

    where beta = alpha_frequency and F is the normalized log FFT magnitude.
    """

    def __init__(
        self,
        input_size=INPUT_SIZE,
        output_size=OUTPUT_SIZE
    ):
        super().__init__()

        self.input_size = input_size
        self.output_size = output_size
        self.frequency_border = output_size - input_size

        if self.frequency_border <= 0:
            raise ValueError(
                "output_size must be larger than input_size."
            )

        # Two global trainable logits.
        # Softmax(logits) -> [alpha_image, alpha_frequency]
        # Initial values are equal: 0.5 and 0.5.
        self.fusion_logits = nn.Parameter(
            torch.zeros(2, dtype=torch.float32)
        )

    def get_weights(self):
        """Return the two differentiable normalized fusion weights."""
        weights = torch.softmax(self.fusion_logits, dim=0)
        alpha_image = weights[0]
        alpha_frequency = weights[1]
        return alpha_image, alpha_frequency

    def extract_frequency(self, x):
        """
        Extract normalized frequency magnitude independently for every
        image and every channel.

        Steps:
            FFT2 -> FFTShift -> Magnitude -> log(1 + magnitude)
            -> per-image/per-channel min-max normalization to [0, 1]
        """

        if x.ndim != 4:
            raise ValueError(
                f"Expected BxCxHxW input, got shape {tuple(x.shape)}"
            )

        fft = torch.fft.fft2(
            x,
            dim=(-2, -1),
            norm="ortho"
        )

        fft = torch.fft.fftshift(
            fft,
            dim=(-2, -1)
        )

        magnitude = torch.abs(fft)
        magnitude = torch.log1p(magnitude)

        min_val = magnitude.amin(
            dim=(-2, -1),
            keepdim=True
        )

        max_val = magnitude.amax(
            dim=(-2, -1),
            keepdim=True
        )

        frequency = (
            magnitude - min_val
        ) / (
            max_val - min_val + 1e-8
        )

        return frequency

    def forward(self, x):

        _, _, height, width = x.shape

        if height != self.input_size or width != self.input_size:
            raise ValueError(
                f"Spatial-frequency module expects "
                f"{self.input_size}x{self.input_size} inputs, "
                f"but received {height}x{width}."
            )

        frequency = self.extract_frequency(x)

        alpha_image, alpha_frequency = self.get_weights()

        # -----------------------------------------------------
        # Spatial part
        # -----------------------------------------------------
        spatial_part = alpha_image * x

        # -----------------------------------------------------
        # Frequency strips used to expand 256x256 -> 300x300
        # -----------------------------------------------------
        border = self.frequency_border

        frequency_right = F.interpolate(
            frequency,
            size=(height, border),
            mode="bilinear",
            align_corners=False
        )

        frequency_bottom = F.interpolate(
            frequency,
            size=(border, width),
            mode="bilinear",
            align_corners=False
        )

        frequency_corner = F.interpolate(
            frequency,
            size=(border, border),
            mode="bilinear",
            align_corners=False
        )

        frequency_right = alpha_frequency * frequency_right
        frequency_bottom = alpha_frequency * frequency_bottom
        frequency_corner = alpha_frequency * frequency_corner

        # Top block: 256 x (256 + 44) = 256 x 300
        top = torch.cat(
            [spatial_part, frequency_right],
            dim=3
        )

        # Bottom block: 44 x (256 + 44) = 44 x 300
        bottom = torch.cat(
            [frequency_bottom, frequency_corner],
            dim=3
        )

        # Final input: 300 x 300
        combined = torch.cat(
            [top, bottom],
            dim=2
        )

        return combined, alpha_image, alpha_frequency


# =====================================================
# MODEL
# =====================================================
class AHCTNet_MultiBranch(nn.Module):

    def __init__(self, num_classes=5):

        super().__init__()

        # =====================================================
        # LEARNABLE SPATIAL-FREQUENCY INPUT MODULE
        # =====================================================
        self.spatial_frequency = LearnableSpatialFrequencyAugmentation(
            input_size=INPUT_SIZE,
            output_size=OUTPUT_SIZE
        )

        # =====================================================
        # BACKBONE
        # =====================================================
        backbone = models.convnext_tiny(
            weights=models.ConvNeXt_Tiny_Weights.DEFAULT
        )

        self.feature_extractor = backbone.features

        feature_dim = 768

        # =====================================================
        # PRE-GATE
        # =====================================================
        self.pre_gate = nn.Sequential(

            nn.AdaptiveAvgPool2d(1),

            nn.Flatten(),

            nn.Linear(feature_dim, 128),

            nn.ReLU(),

            nn.Dropout(0.2),

            nn.Linear(128, 1),

            nn.Sigmoid()
        )

        # =====================================================
        # BRANCH 1
        # CNN LOCAL
        # =====================================================
        self.branch1 = nn.Sequential(

            nn.Conv2d(
                feature_dim,
                256,
                3,
                padding=1
            ),

            nn.BatchNorm2d(256),

            nn.ReLU(),

            nn.Dropout2d(0.2),

            nn.AdaptiveAvgPool2d(1)
        )

        self.branch1_fc = nn.Linear(256, 128)

        # =====================================================
        # BRANCH 2
        # DEPTHWISE CNN
        # =====================================================
        self.branch2 = nn.Sequential(

            nn.Conv2d(
                feature_dim,
                feature_dim,
                kernel_size=3,
                padding=1,
                groups=feature_dim
            ),

            nn.Conv2d(
                feature_dim,
                256,
                kernel_size=1
            ),

            nn.BatchNorm2d(256),

            nn.ReLU(),

            nn.Dropout2d(0.2),

            nn.AdaptiveAvgPool2d(1)
        )

        self.branch2_fc = nn.Linear(256, 128)

        # =====================================================
        # BRANCH 3
        # TRANSFORMER GLOBAL
        # =====================================================
        self.patch3 = nn.Conv2d(
            feature_dim,
            128,
            1
        )

        self.transformer3 = nn.TransformerEncoder(

            nn.TransformerEncoderLayer(
                d_model=128,
                nhead=4,
                dim_feedforward=512,
                dropout=0.1,
                batch_first=True
            ),

            num_layers=2
        )

        # =====================================================
        # BRANCH 4
        # MULTI-SCALE TRANSFORMER
        # =====================================================
        self.patch4 = nn.Conv2d(
            feature_dim,
            128,
            3,
            padding=1
        )

        self.transformer4 = nn.TransformerEncoder(

            nn.TransformerEncoderLayer(
                d_model=128,
                nhead=8,
                dim_feedforward=512,
                dropout=0.1,
                batch_first=True
            ),

            num_layers=2
        )

        # =====================================================
        # ROUTER
        # =====================================================
        self.router = nn.Sequential(

            nn.Linear(feature_dim, 128),

            nn.ReLU(),

            nn.Dropout(0.2),

            nn.Linear(128, 4)
        )

        # =====================================================
        # CLASSIFIER
        # =====================================================
        self.fc = nn.Sequential(

            nn.Linear(128, 64),

            nn.ReLU(),

            nn.Dropout(0.3),

            nn.Linear(64, num_classes)
        )

    # =====================================================
    # FORWARD
    # =====================================================
    def forward(self, x):

        # =====================================================
        # LEARNABLE SPATIAL-FREQUENCY AUGMENTATION
        # 256x256 -> 300x300
        # =====================================================
        x, alpha_image, alpha_frequency = self.spatial_frequency(x)

        feat = self.feature_extractor(x)

        # =====================================================
        # PRE-GATE
        # =====================================================
        g1 = self.pre_gate(feat)

        feat = feat * (
            0.5 + 0.5 * g1.unsqueeze(-1).unsqueeze(-1)
        )

        # =====================================================
        # GLOBAL FEATURE
        # =====================================================
        global_feat = F.adaptive_avg_pool2d(
            feat,
            1
        ).flatten(1)

        # =====================================================
        # BRANCH 1
        # =====================================================
        b1 = self.branch1(feat)

        b1 = b1.view(x.size(0), -1)

        b1 = self.branch1_fc(b1)

        # =====================================================
        # BRANCH 2
        # =====================================================
        b2 = self.branch2(feat)

        b2 = b2.view(x.size(0), -1)

        b2 = self.branch2_fc(b2)

        # =====================================================
        # BRANCH 3
        # =====================================================
        t3 = self.patch3(feat)

        t3 = t3.flatten(2).permute(0,2,1)

        t3 = self.transformer3(t3)

        b3 = t3.mean(dim=1)

        # =====================================================
        # BRANCH 4
        # =====================================================
        t4 = self.patch4(feat)

        t4 = t4.flatten(2).permute(0,2,1)

        t4 = self.transformer4(t4)

        b4 = t4.mean(dim=1)

        # =====================================================
        # STACK FEATURES
        # =====================================================
        features = torch.stack(
            [b1, b2, b3, b4],
            dim=1
        )

        # =====================================================
        # ROUTER
        # =====================================================
        scores = self.router(global_feat)

        weights = torch.softmax(
            scores,
            dim=1
        )

        # =====================================================
        # TOP-2 GATING
        # =====================================================
        topk_vals, topk_idx = torch.topk(
            weights,
            k=2,
            dim=1
        )

        mask = torch.zeros_like(weights)

        mask.scatter_(1, topk_idx, 1)

        sparse_weights = weights * mask

        sparse_weights = sparse_weights / (
            sparse_weights.sum(
                dim=1,
                keepdim=True
            ) + 1e-8
        )

        # =====================================================
        # FUSION
        # =====================================================
        fused = (
            features *
            sparse_weights.unsqueeze(-1)
        ).sum(dim=1)

        # =====================================================
        # CLASSIFICATION
        # =====================================================
        out = self.fc(fused)

        return (
            out,
            sparse_weights,
            g1,
            alpha_image,
            alpha_frequency
        )

    def get_spatial_frequency_weights(self):
        """Return current fusion weights as Python floats."""
        alpha_image, alpha_frequency = (
            self.spatial_frequency.get_weights()
        )

        return (
            alpha_image.detach().cpu().item(),
            alpha_frequency.detach().cpu().item()
        )


# =====================================================
# TRAIN
# =====================================================
def train_epoch(
    model,
    loader,
    optimizer,
    criterion
):

    model.train()

    losses = []

    for x, y in loader:

        x = x.to(device)
        y = y.to(device)

        out, gates, g1, alpha_image, alpha_frequency = model(x)

        loss = criterion(out, y)

        optimizer.zero_grad()

        loss.backward()

        optimizer.step()

        losses.append(loss.item())

    return np.mean(losses)


# =====================================================
# EVALUATION
# =====================================================
def evaluate(
    model,
    loader,
    num_classes=5
):

    model.eval()

    preds = []
    trues = []
    probs = []
    all_gates = []

    with torch.no_grad():

        for x, y in loader:

            x = x.to(device)

            out, gates, g1, alpha_image, alpha_frequency = model(x)

            p = torch.softmax(out, dim=1)

            preds.extend(
                torch.argmax(p,1).cpu().numpy()
            )

            probs.extend(
                p.cpu().numpy()
            )

            trues.extend(
                y.numpy()
            )

            all_gates.extend(
                gates.cpu().numpy()
            )

    probs = np.array(probs)

    metrics = {

        "Accuracy":
            accuracy_score(trues, preds),

        "Precision":
            precision_score(
                trues,
                preds,
                average='weighted',
                zero_division=0
            ),

        "Recall":
            recall_score(
                trues,
                preds,
                average='weighted',
                zero_division=0
            ),

        "F1":
            f1_score(
                trues,
                preds,
                average='weighted'
            ),

        "Balanced_ACC":
            balanced_accuracy_score(
                trues,
                preds
            ),

        "MCC":
            matthews_corrcoef(
                trues,
                preds
            ),

        "Kappa":
            cohen_kappa_score(
                trues,
                preds
            )
    }

    try:

        metrics["AUC"] = roc_auc_score(
            trues,
            probs,
            multi_class='ovr'
        )

    except:

        metrics["AUC"] = 0

    return (
        metrics,
        trues,
        preds,
        probs,
        np.array(all_gates)
    )


# =====================================================
# PRINT EPOCH METRICS
# =====================================================
def print_epoch_metrics(
    fold,
    epoch,
    train_loss,
    metrics,
    lr,
    alpha_image,
    alpha_frequency
):

    print(
        f"Fold {fold} | "
        f"Epoch {epoch} | "
        f"LR {lr:.8f} | "
        f"Train_Loss {train_loss:.4f} | "
        f"Accuracy {metrics['Accuracy']:.4f} | "
        f"Precision {metrics['Precision']:.4f} | "
        f"Recall {metrics['Recall']:.4f} | "
        f"F1 {metrics['F1']:.4f} | "
        f"Balanced_ACC {metrics['Balanced_ACC']:.4f} | "
        f"MCC {metrics['MCC']:.4f} | "
        f"Kappa {metrics['Kappa']:.4f} | "
        f"AUC {metrics['AUC']:.4f} | "
        f"Alpha_Image {alpha_image:.4f} | "
        f"Alpha_Frequency {alpha_frequency:.4f}"
    )


# =====================================================
# PLOTS
# =====================================================
def plot_confusion(y_true, y_pred):

    cm = confusion_matrix(y_true, y_pred)

    sns.heatmap(cm, annot=True, fmt='d')

    plt.title("Confusion Matrix")

    plt.show()


def plot_roc(y_true, probs, n_classes):

    for i in range(n_classes):

        fpr, tpr, _ = roc_curve(
            np.array(y_true)==i,
            probs[:,i]
        )

        plt.plot(
            fpr,
            tpr,
            label=f"class {i}"
        )

    plt.legend()

    plt.title("ROC Curve")

    plt.show()


def plot_gate_usage(gates):

    mean_usage = gates.mean(axis=0)

    plt.figure(figsize=(6,4))

    plt.bar(
        ["B1","B2","B3","B4"],
        mean_usage
    )

    plt.title("Average Branch Usage")

    plt.ylabel("Average Gate Weight")

    plt.show()


def plot_spatial_frequency_weights(
    alpha_image_history,
    alpha_frequency_history,
    fold
):

    epochs = np.arange(
        1,
        len(alpha_image_history) + 1
    )

    plt.figure(figsize=(7,4))

    plt.plot(
        epochs,
        alpha_image_history,
        marker="o",
        label="Alpha Image"
    )

    plt.plot(
        epochs,
        alpha_frequency_history,
        marker="o",
        label="Alpha Frequency"
    )

    plt.ylim(0, 1)
    plt.xlabel("Epoch")
    plt.ylabel("Learnable Weight")
    plt.title(
        f"Spatial-Frequency Weights - Fold {fold}"
    )
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()


# =====================================================
# BRANCH ANALYSIS
# =====================================================
def analyze_branch_usage_per_class(
    y_true,
    gates,
    class_names=None
):

    y_true = np.array(y_true)

    gates = np.array(gates)

    n_classes = len(np.unique(y_true))

    if class_names is None:

        class_names = [
            f"Class {i}"
            for i in range(n_classes)
        ]

    print(
        "\n========== BRANCH USAGE PER CLASS ==========\n"
    )

    for c in range(n_classes):

        idx = np.where(y_true == c)[0]

        class_gates = gates[idx]

        mean_usage = class_gates.mean(axis=0)

        print(f"\n{class_names[c]}")

        print(
            f"Branch1 CNN         : "
            f"{mean_usage[0]:.4f}"
        )

        print(
            f"Branch2 DW-CNN      : "
            f"{mean_usage[1]:.4f}"
        )

        print(
            f"Branch3 Transformer : "
            f"{mean_usage[2]:.4f}"
        )

        print(
            f"Branch4 MS-Trans    : "
            f"{mean_usage[3]:.4f}"
        )

        most_used = np.argmax(mean_usage)

        print(
            f"Most Used Branch --> "
            f"B{most_used+1}"
        )

        print("-"*50)


def plot_branch_heatmap(
    y_true,
    gates,
    class_names=None
):

    y_true = np.array(y_true)

    gates = np.array(gates)

    n_classes = len(np.unique(y_true))

    if class_names is None:

        class_names = [
            f"Class {i}"
            for i in range(n_classes)
        ]

    heatmap_data = []

    for c in range(n_classes):

        idx = np.where(y_true == c)[0]

        class_gates = gates[idx]

        mean_usage = class_gates.mean(axis=0)

        heatmap_data.append(mean_usage)

    heatmap_data = np.array(heatmap_data)

    plt.figure(figsize=(8,5))

    sns.heatmap(
        heatmap_data,
        annot=True,
        cmap="viridis",

        xticklabels=[
            "B1 CNN",
            "B2 DW-CNN",
            "B3 Transformer",
            "B4 MS-Transformer"
        ],

        yticklabels=class_names
    )

    plt.title("Branch Usage Per Class")

    plt.xlabel("Branches")

    plt.ylabel("Classes")

    plt.show()


# =====================================================
# MAIN
# =====================================================
root = "/kaggle/input/datasets/vibhingupta028/liver-histopathology-fibrosis-ultrasound-images/Dataset/Dataset"

samples = load_dataset(root)

labels = [s[1] for s in samples]

num_classes = len(set(labels))

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

all_fold_metrics = []
all_epoch_metrics = []

global_best_acc = -1
global_best_info = None

for fold, (train_idx, val_idx) in enumerate(
    skf.split(samples, labels)
):

    print(
        f"\n========== FOLD {fold+1} ==========\n"
    )

    train_samples = [
        samples[i]
        for i in train_idx
    ]

    val_samples = [
        samples[i]
        for i in val_idx
    ]

    train_loader = DataLoader(
        LiverDataset(train_samples, train_tf),
        batch_size=16,
        shuffle=True
    )

    val_loader = DataLoader(
        LiverDataset(val_samples, val_tf),
        batch_size=16
    )

    model = AHCTNet_MultiBranch(
        num_classes=num_classes
    ).to(device)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=1e-4,
        weight_decay=1e-4
    )

    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=15
    )

    criterion = FocalLoss()

    best_acc = -1
    best_epoch = 0
    best_fold_metrics = None
    best_fold_model_path = os.path.join(
        OUTPUT_DIR,
        f"best_fold_{fold+1}.pth"
    )

    train_losses = []
    val_accs = []
    alpha_image_history = []
    alpha_frequency_history = []

    for epoch in range(20):

        loss = train_epoch(
            model,
            train_loader,
            optimizer,
            criterion
        )

        metrics, y_true, y_pred, probs, gates = evaluate(
            model,
            val_loader,
            num_classes
        )

        current_lr = optimizer.param_groups[0]["lr"]

        alpha_image, alpha_frequency = (
            model.get_spatial_frequency_weights()
        )

        scheduler.step()

        train_losses.append(loss)

        val_accs.append(metrics["Accuracy"])

        alpha_image_history.append(alpha_image)
        alpha_frequency_history.append(alpha_frequency)

        epoch_result = {
            "Fold": fold + 1,
            "Epoch": epoch + 1,
            "Train_Loss": loss,
            "Learning_Rate": current_lr,
            "Alpha_Image": alpha_image,
            "Alpha_Frequency": alpha_frequency,
            **metrics
        }

        all_epoch_metrics.append(epoch_result)

        pd.DataFrame(all_epoch_metrics).to_csv(
            EPOCH_METRICS_CSV,
            index=False
        )

        print_epoch_metrics(
            fold + 1,
            epoch + 1,
            loss,
            metrics,
            current_lr,
            alpha_image,
            alpha_frequency
        )

        if metrics["Accuracy"] > best_acc:

            best_acc = metrics["Accuracy"]
            best_epoch = epoch + 1
            best_fold_metrics = metrics.copy()
            best_fold_metrics["Best_Epoch"] = best_epoch
            best_fold_metrics["Alpha_Image"] = alpha_image
            best_fold_metrics["Alpha_Frequency"] = alpha_frequency

            torch.save(
                model.state_dict(),
                best_fold_model_path
            )

        if metrics["Accuracy"] > global_best_acc:

            global_best_acc = metrics["Accuracy"]

            global_best_info = {
                "Best_Fold": fold + 1,
                "Best_Epoch": epoch + 1,
                "Train_Loss": loss,
                "Learning_Rate": current_lr,
                "Alpha_Image": alpha_image,
                "Alpha_Frequency": alpha_frequency,
                **metrics
            }

            torch.save(
                model.state_dict(),
                BEST_OVERALL_MODEL_PATH
            )

            pd.DataFrame([global_best_info]).to_csv(
                BEST_OVERALL_INFO_CSV,
                index=False
            )

    all_fold_metrics.append(best_fold_metrics)

    print(
        f"\nBest model for Fold {fold+1}: "
        f"Epoch {best_epoch} | "
        f"Accuracy {best_fold_metrics['Accuracy']:.4f} | "
        f"Alpha_Image {best_fold_metrics['Alpha_Image']:.4f} | "
        f"Alpha_Frequency {best_fold_metrics['Alpha_Frequency']:.4f}"
    )

    model.load_state_dict(
        torch.load(
            best_fold_model_path,
            map_location=device
        )
    )

    metrics, y_true, y_pred, probs, gates = evaluate(
        model,
        val_loader,
        num_classes
    )

    # =====================================================
    # PLOTS
    # =====================================================
    plt.plot(train_losses)
    plt.title("Loss")
    plt.show()

    plt.plot(val_accs)
    plt.title("Accuracy")
    plt.show()

    plot_spatial_frequency_weights(
        alpha_image_history,
        alpha_frequency_history,
        fold + 1
    )

    plot_confusion(y_true, y_pred)

    plot_roc(y_true, probs, num_classes)

    plot_gate_usage(gates)

    # =====================================================
    # CLASS NAMES
    # =====================================================
    class_names = [
        "Class0",
        "Class1",
        "Class2",
        "Class3",
        "Class4"
    ]

    # =====================================================
    # BRANCH ANALYSIS
    # =====================================================
    analyze_branch_usage_per_class(
        y_true,
        gates,
        class_names
    )

    # =====================================================
    # HEATMAP
    # =====================================================
    plot_branch_heatmap(
        y_true,
        gates,
        class_names
    )


# =====================================================
# FINAL RESULTS
# =====================================================
print(
    "\n========== FINAL RESULTS ==========\n"
)

for k in all_fold_metrics[0].keys():

    vals = [m[k] for m in all_fold_metrics]

    print(k)

    print("Mean:", np.mean(vals))

    print("Std:", np.std(vals))
    print("-"*40) 
```
